```python
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
model = load_model('model.h5')
```

```python
with open('gender_encoder.pkl','rb') as file:
    gender_label_encoder = pickle.load(file)
```

```python
with open('geography_encoder.pkl','rb') as file:
    geo_one_hot_encoder = pickle.load(file)
```

```python
input_data = {
    'CreditScore':600,
    'Geography':'France',
    'Gender':'Male',
    'Age':40,
    'Tenure':3,
    'Balance':60000,
    'NumOfProducts':2,
    'HasCrCard':1,
    'IsActiveMember':1,
    'EstimatedSalary':50000

}
```

```python
test_df = pd.DataFrame([input_data])
```

```python
test_df
```

```python
geo_df = pd.DataFrame(geo_one_hot_encoder.transform(test_df[['Geography']]).toarray(),columns = geo_one_hot_encoder.get_feature_names_out())
```

```python
geo_df
```

```python
test_df = pd.concat([test_df,geo_df],axis=1)
```

```python
test_df
```

```python
test_df.drop(['Geography'],axis=1,inplace=True)
```

```python
test_df['Gender'] = gender_label_encoder.transform(test_df['Gender'])
```

```python
test_df
```

```python
with open('scaler.pkl','rb') as file:
    scaler = pickle.load(file)
```

```python
test_df = scaler.transform(test_df)
```

```python
test_df
```

```python
output = model.predict(test_df)
```

```python
final_output = output[0][0]
```

```python
if final_output > 0.5:
    print("The customer will churn")
else:
    print("The customer will not churn")
```
