```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder,LabelEncoder,StandardScaler
from sklearn.compose import ColumnTransformer
import pickle
```

```python
df = pd.read_csv('Churn_Modelling.csv')
```

```python
df.head(10)
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df.drop(['RowNumber','CustomerId','Surname'],axis=1,inplace=True)
```

```python
df.head(10)
```

```python
one_hot_encoder = OneHotEncoder()
```

```python
transformed_geo = pd.DataFrame(one_hot_encoder.fit_transform(df[['Geography']]).toarray(),columns = one_hot_encoder.get_feature_names_out())
```

```python
transformed_geo
```

```python
df = pd.concat([df,transformed_geo],axis=1)
```

```python
df
```

```python
df.isnull().sum()
```

```python
df.drop(['Geography'],axis=1,inplace=True)
```

```python
df.head(10)
```

```python
encoder_label = LabelEncoder()
```

```python
df['Gender'] = encoder_label.fit_transform(df['Gender'])
```

```python
df.head(10)
```

```python
with open('gender_encoder.pkl','wb') as file:
    pickle.dump(encoder_label,file)
```

```python
with open('geography_encoder.pkl','wb') as file:
    pickle.dump(one_hot_encoder,file)
```

```python
x = df.drop(['Exited'],axis=1)
y = df['Exited']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
scaler = StandardScaler()
```

```python
x_train_scaled = scaler.fit_transform(x_train)
```

```python
x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled
```

```python
with open('scaler.pkl','wb') as file:
    pickle.dump(scaler,file)
```

```python
x_train_scaled.shape
```

```python
import tensorflow as tf
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
import datetime
```

```python
#Building ANN Model
```

```python
model = Sequential([Dense(64,activation='relu',input_shape = (x_train.shape[1],)),
                   Dense(32,activation='relu'),
                   Dense(1,activation='sigmoid')])
```

```python
model.summary()
```

```python
opt = tf.keras.optimizers.Adam(learning_rate = 0.01)
```

```python
model.compile(optimizer=opt,loss='binary_crossentropy',metrics=['accuracy'])
```

```python
#Setup Tensorboard
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
log_dir = "logs/fit/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tensorflow_callback = TensorBoard(log_dir=log_dir,histogram_freq=1)
```

```python
#Setup Early Stopping
early_stopping_callback = EarlyStopping(monitor='val_loss',patience=5,restore_best_weights=True)
```

```python
history = model.fit(x_train_scaled,y_train,validation_data=(x_test_scaled,y_test),epochs=100,callbacks=[tensorflow_callback,early_stopping_callback])
```

```python
model.save('model.h5')
```
